# CineMatch AI – Exploratory Data Analysis (EDA)
**Dataset:** MovieLens 100K (100,000 ratings from 943 users on 1,682 movies)

This notebook provides exploratory analysis into:
1. Rating distributions and user scoring tendencies
2. Movie popularity and the long-tail phenomenon
3. Genre popularity vs. quality analysis
4. User demographics (age, gender, occupation)
5. Release year trends across cinema history

```python
import sys
from pathlib import Path
sys.path.append(str(Path("..").resolve()))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.preprocessing import load_processed_data

plt.style.use('seaborn-v0_8-darkgrid')
ratings, movies, users = load_processed_data()
print(f"Ratings shape: {ratings.shape}")
print(f"Movies shape:  {movies.shape}")
print(f"Users shape:   {users.shape}")
```

## 1. Rating Distribution
Examining the overall distribution of rating values (1.0 to 5.0 stars).

```python
plt.figure(figsize=(8, 4.5))
sns.countplot(data=ratings, x='rating', palette='viridis')
plt.title('Distribution of Movie Ratings (1 to 5 Stars)', fontsize=14, fontweight='bold')
plt.xlabel('Rating (Stars)')
plt.ylabel('Count')
plt.show()
```

## 2. Top 10 Most-Rated Movies
Which movies garnered the highest volume of community reviews?

```python
top_movies = ratings.groupby('movie_id').size().reset_index(name='count')
top_movies = top_movies.merge(movies[['movie_id', 'clean_title', 'year']], on='movie_id')
top_10 = top_movies.nlargest(10, 'count')

plt.figure(figsize=(10, 5))
sns.barplot(data=top_10, y='clean_title', x='count', palette='plasma')
plt.title('Top 10 Most-Rated Movies', fontsize=14, fontweight='bold')
plt.xlabel('Rating Count')
plt.ylabel('Title')
plt.show()
```

## 3. Genre Distribution Analysis
Comparing the volume of titles per genre across the 18 catalog categories.

```python
genre_cols = [c for c in movies.columns if c not in ['movie_id', 'title', 'clean_title', 'release_date', 'imdb_url', 'year', 'genres']]
genre_counts = movies[genre_cols].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(x=genre_counts.values, y=genre_counts.index, palette='mako')
plt.title('Movie Catalog Count by Genre', fontsize=14, fontweight='bold')
plt.xlabel('Number of Movies')
plt.ylabel('Genre')
plt.show()
```

## 4. User Activity: Ratings per User Distribution

```python
user_ratings_count = ratings.groupby('user_id').size()
plt.figure(figsize=(9, 4.5))
sns.histplot(user_ratings_count, bins=40, kde=True, color='purple')
plt.title('Distribution of Ratings per User (Activity)', fontsize=14, fontweight='bold')
plt.xlabel('Number of Ratings per User')
plt.ylabel('User Count')
plt.show()
```
